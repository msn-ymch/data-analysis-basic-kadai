```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 30)
y = 2 + 2 * np.sin(2 * x)

fig, ax = plt.subplots()

ax.plot(x, y, linewidth=2.0)

ax.set(
    xlim=(0, 10), xticks=np.arange(0, 10),
    ylim=(-1, 5), yticks=np.arange(-1, 6),
    xlabel="x",
    ylabel="y",
    title="Simple Plot",
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import japanize_matplotlib

fig, axes = plt.subplots(2, 3,
    tight_layout=True)

# 折れ線グラフ
axes[0, 0].plot([0, 1, 2, 3],
    [1, 2, -5, 2])

# sin関数
axes[0, 1].plot(
      np.linspace(0, 10, 100),
      np.sin(np.linspace(0, 10, 100)),
      linewidth=2.0
    )

# ヒストグラム
axes[0, 2].hist(np.random.normal(15, 5, 2000))

# 散布図
axes[1, 0].scatter(
    np.random.normal(15, 5, 2000),
    np.random.normal(15, 5, 2000)
)

plt.show()
```
